```python
file_names = [
    "text_spok.txt",
    "text_blog.txt",
    "text_tvm.txt",
    "text_news.txt",
    "text_fic.txt",
    "text_web.txt",
    "text_acad.txt",
    "text_mag.txt"
]

all_texts = {}

for file_name in file_names:
    file_path = "/content/" + file_name

    with open(file_path, "r", encoding="utf-8") as file:
        all_texts[file_name] = file.read()

print("Files loaded:", len(all_texts))
```

```python
all_lines = []

for file_name, text in all_texts.items():
    lines = text.splitlines()

    for line in lines:
        line = line.strip()

        if len(line) > 20:
            all_lines.append(line)

print("Total usable lines:", len(all_lines))
```

```python
for i in range(8):
    print(all_lines[i])
    print("-" * 60)
```
