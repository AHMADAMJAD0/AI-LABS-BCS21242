```python
import random

# Possible values for each hyperparameter
SEARCH_SPACE = {
    "conv1_filters": [16, 32, 64],
    "conv2_filters": [32, 64, 128],
    "dense_neurons": [64, 128, 256],
    "dropout": [0.2, 0.3, 0.4, 0.5],
    "learning_rate": [0.0001, 0.0005, 0.001, 0.005]
}


# Create one random chromosome
def create_chromosome():
    chromosome = {
        "conv1_filters": random.choice(SEARCH_SPACE["conv1_filters"]),
        "conv2_filters": random.choice(SEARCH_SPACE["conv2_filters"]),
        "dense_neurons": random.choice(SEARCH_SPACE["dense_neurons"]),
        "dropout": random.choice(SEARCH_SPACE["dropout"]),
        "learning_rate": random.choice(SEARCH_SPACE["learning_rate"])
    }

    return chromosome


# Test it
chromosome = create_chromosome()

print(chromosome)
```

```python
# Create a population of chromosomes
def create_population(population_size):
    population = []

    for _ in range(population_size):
        chromosome = create_chromosome()
        population.append(chromosome)

    return population


# Create 6 different CNN configurations
population = create_population(6)

for i, chromosome in enumerate(population):
    print(f"Chromosome {i+1}:")
    print(chromosome)
    print()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class GeneticCNN(nn.Module):
    def __init__(self, chromosome):
        super().__init__()

        conv1_filters = chromosome["conv1_filters"]
        conv2_filters = chromosome["conv2_filters"]
        dense_neurons = chromosome["dense_neurons"]
        dropout = chromosome["dropout"]

        self.conv1 = nn.Conv2d(1, conv1_filters, kernel_size=3)
        self.conv2 = nn.Conv2d(conv1_filters, conv2_filters, kernel_size=3)

        self.pool = nn.MaxPool2d(2, 2)

        self.fc1 = nn.Linear(
            conv2_filters * 5 * 5,
            dense_neurons
        )

        self.dropout = nn.Dropout(dropout)

        self.fc2 = nn.Linear(dense_neurons, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)

        return x


def build_model(chromosome):
    return GeneticCNN(chromosome)
```

```python
from torch.utils.data import DataLoader

def evaluate_chromosome(chromosome, epochs=3):

    # Build model
    model = build_model(chromosome)

    # Loss function
    criterion = nn.CrossEntropyLoss()

    # Optimizer
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=chromosome["learning_rate"]
    )

    # DataLoaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=32,
        shuffle=True
    )

    val_loader = DataLoader(
        val_dataset,
        batch_size=32,
        shuffle=False
    )

    # -------------------------
    # TRAINING
    # -------------------------
    model.train()

    for epoch in range(epochs):

        for images, labels in train_loader:

            optimizer.zero_grad()

            outputs = model(images)

            loss = criterion(outputs, labels)

            loss.backward()

            optimizer.step()

    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in val_loader:

            outputs = model(images)

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)

            correct += (predicted == labels).sum().item()

    accuracy = correct / total

    return accuracy
```

```python
def tournament_selection(population, fitness_scores, tournament_size=3):

    # Randomly choose some chromosome indexes
    selected_indexes = random.sample(
        range(len(population)),
        tournament_size
    )

    # Find the best one among them
    best_index = max(
        selected_indexes,
        key=lambda i: fitness_scores[i]
    )

    return population[best_index]
```

```python
def crossover(parent1, parent2):

    child = {}

    for gene in parent1.keys():

        # Randomly inherit gene from either parent
        if random.random() < 0.5:
            child[gene] = parent1[gene]
        else:
            child[gene] = parent2[gene]

    return child
```

```python
parent1 = tournament_selection(
    population,
    fitness_scores
)

parent2 = tournament_selection(
    population,
    fitness_scores
)

child = crossover(parent1, parent2)

print("Parent 1:")
print(parent1)

print("\nParent 2:")
print(parent2)

print("\nChild:")
print(child)
```

```python
def mutate(chromosome, mutation_rate=0.2):

    mutated = chromosome.copy()

    for gene in mutated.keys():

        if random.random() < mutation_rate:

            mutated[gene] = random.choice(
                SEARCH_SPACE[gene]
            )

    return mutated
```

```python
mutated_child = mutate(child)

print("Before Mutation:")
print(child)

print("\nAfter Mutation:")
print(mutated_child)
```

```python
def create_new_generation(population, fitness_scores, population_size):

    new_population = []

    while len(new_population) < population_size:

        # Select two parents
        parent1 = tournament_selection(
            population,
            fitness_scores
        )

        parent2 = tournament_selection(
            population,
            fitness_scores
        )

        # Crossover
        child = crossover(parent1, parent2)

        # Mutation
        child = mutate(child)

        # Add child to new population
        new_population.append(child)

    return new_population
```

```python
new_population = create_new_generation(
    population,
    fitness_scores,
    population_size=6
)

for i, chromosome in enumerate(new_population):
    print(f"New Chromosome {i+1}:")
    print(chromosome)
    print()
```

```python
# Number of generations
NUM_GENERATIONS = 5

# Start with the current population
population = new_population

best_history = []

for generation in range(NUM_GENERATIONS):

    print(f"\n========== Generation {generation + 1} ==========")

    fitness_scores = []

    # 1. Evaluate every chromosome
    for i, chromosome in enumerate(population):

        fitness = evaluate_chromosome(
            chromosome,
            epochs=3
        )

        fitness_scores.append(fitness)

        print(
            f"Chromosome {i+1} "
            f"Fitness: {fitness:.4f}"
        )

    # 2. Find best chromosome of this generation
    best_index = fitness_scores.index(
        max(fitness_scores)
    )

    best_chromosome = population[best_index]
    best_fitness = fitness_scores[best_index]

    best_history.append(best_fitness)

    print("\nBest chromosome:")
    print(best_chromosome)

    print(
        f"Best fitness: {best_fitness:.4f}"
    )

    # 3. Create next generation
    population = create_new_generation(
        population,
        fitness_scores,
        population_size=6
    )
```

```python
import matplotlib.pyplot as plt

plt.plot(
    range(1, len(best_history) + 1),
    best_history,
    marker='o'
)

plt.xlabel("Generation")
plt.ylabel("Best Validation Accuracy")
plt.title("Genetic Algorithm Optimization")

plt.xticks(range(1, len(best_history) + 1))
plt.grid()

plt.show()
```
