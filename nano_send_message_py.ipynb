```python
import pywhatkit
import datetime

phone_number = "+923017498063"
message = "I love you ❤️"

now = datetime.datetime.now()

hour = now.hour
minute = now.minute + 2

pywhatkit.sendwhatmsg(
    phone_number,
    message,
    hour,
    minute
)
```

```python
import pywhatkit
from datetime import datetime, timedelta

phone_number = "+923017498063"
message = "I love you ❤️"

send_time = datetime.now() + timedelta(minutes=2)

pywhatkit.sendwhatmsg(
    phone_number,
    message,
    send_time.hour,
    send_time.minute
)
```
