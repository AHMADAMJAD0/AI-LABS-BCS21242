```python
file_path = "text_news.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Total characters:", len(text))
print(text[:1000])
```

```python
text = text.lower()

print(text[:1000])
```

```python
import re
```

```python
text = re.sub(r"<[^>]+>", " ", text)
```

```python
text = re.sub(r"@@\d+", " ", text)
```

```python
text = re.sub(r"@", " ", text)
```

```python
text = re.sub(r"http\S+|www\.\S+", " ", text)
```

```python
text = re.sub(r"[^a-z\s]", " ", text)
```

```python
text = re.sub(r"\s+", " ", text).strip()
```

```python
words = text.split()

print(words[:50])
print("Total words:", len(words))
```

```python
stop_words = {
    "a", "an", "the",
    "and", "or", "but",
    "is", "am", "are", "was", "were",
    "be", "been", "being",
    "of", "to", "in", "on", "at",
    "for", "from", "with", "by",
    "this", "that", "these", "those",
    "it", "its",
    "he", "she", "they", "we", "you", "i",
    "his", "her", "their", "our", "your",
    "as", "if", "then", "than",
    "not", "no"
}
```

```python
clean_words = []

for word in words:
    if word not in stop_words:
        clean_words.append(word)
```

```python
cleaned_paragraphs = [word for word in words if word not in stop_words]
```

```python
clean_text = " ".join(cleaned_paragraphs)

print(clean_text[:5000])
```

```python
# Join paragraphs with a blank line between them
clean_text = "\n\n".join(cleaned_paragraphs)

print(clean_text[:3000])
```
