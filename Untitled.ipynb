```python
import random

# Possible values for each hyperparameter
SEARCH_SPACE = {
    "conv1_filters": [16, 32, 64],
    "conv2_filters": [32, 64, 128],
    "dense_neurons": [64, 128, 256],
    "dropout": [0.2, 0.3, 0.4, 0.5],
    "learning_rate": [0.0001, 0.0005, 0.001, 0.005]
}


# Create one random chromosome
def create_chromosome():
    chromosome = {
        "conv1_filters": random.choice(SEARCH_SPACE["conv1_filters"]),
        "conv2_filters": random.choice(SEARCH_SPACE["conv2_filters"]),
        "dense_neurons": random.choice(SEARCH_SPACE["dense_neurons"]),
        "dropout": random.choice(SEARCH_SPACE["dropout"]),
        "learning_rate": random.choice(SEARCH_SPACE["learning_rate"])
    }

    return chromosome


# Test it
chromosome = create_chromosome()

print(chromosome)
```

```python
# Create a population of chromosomes
def create_population(population_size):
    population = []

    for _ in range(population_size):
        chromosome = create_chromosome()
        population.append(chromosome)

    return population


# Create 6 different CNN configurations
population = create_population(6)

for i, chromosome in enumerate(population):
    print(f"Chromosome {i+1}:")
    print(chromosome)
    print()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class GeneticCNN(nn.Module):
    def __init__(self, chromosome):
        super().__init__()

        conv1_filters = chromosome["conv1_filters"]
        conv2_filters = chromosome["conv2_filters"]
        dense_neurons = chromosome["dense_neurons"]
        dropout = chromosome["dropout"]

        self.conv1 = nn.Conv2d(1, conv1_filters, kernel_size=3)
        self.conv2 = nn.Conv2d(conv1_filters, conv2_filters, kernel_size=3)

        self.pool = nn.MaxPool2d(2, 2)

        self.fc1 = nn.Linear(
            conv2_filters * 5 * 5,
            dense_neurons
        )

        self.dropout = nn.Dropout(dropout)

        self.fc2 = nn.Linear(dense_neurons, 10)

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)

        return x


def build_model(chromosome):
    return GeneticCNN(chromosome)
```

```python
model = build_model(chromosome)

print(model)
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=chromosome["learning_rate"]
)

print(optimizer)
```
